```python
"""Library data generation script for circulation dataset creation."""

import csv
import random
from datetime import date, timedelta
```

```python
# Configuration for the dataset
TOTAL_ROWS = 2000
FILENAME = "Library_Circulation_Data.csv"
```

```python
branches = ["Main Branch", "North Branch", "South Branch"]
patron_types = ["Adult", "Student", "Senior", "Teacher"]
categories = {
    "Fiction": {"cost": [15.00, 20.00, 22.00, 24.00, 28.00], "duration": 14},
    "Non-Fiction": {"cost": [25.00, 30.00, 35.00, 40.00, 45.00], "duration": 14},
    "Audiobook": {"cost": [15.00, 18.00, 20.00, 25.00], "duration": 14},
    "Reference": {"cost": [85.00, 95.00, 120.00, 150.00], "duration": 14},
    "Graphic Novel": {"cost": [14.00, 15.00, 16.00, 18.00], "duration": 14},
    "DVD": {"cost": [15.00, 18.00, 20.00], "duration": 7}
}
```

```python
# Generate random dates
start_date = date(2025, 1, 1)
end_date = date(2025, 12, 1)
date_range = (end_date - start_date).days
```

```python
with open(FILENAME, mode="w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    # Write header
    writer.writerow([
        "TransactionID", "CheckoutDate", "DueDate", 
        "ReturnDate", "Branch", "PatronType", 
        "ItemCategory", "FineAmount", "ReplacementCost"
    ])

    for i in range(TOTAL_ROWS):
        transaction_id = 10001 + i

        # Determine dates
        random_days = random.randint(0, date_range)
        checkout_date = start_date + timedelta(days=random_days)

        category_name = random.choice(list(categories.keys()))
        duration = categories[category_name]["duration"]
        due_date = checkout_date + timedelta(days=duration)

        # Simulate returns (some early, some late)
        return_offset = random.randint(-5, 10)
        return_date = checkout_date + timedelta(days=(duration + return_offset))

        # Calculate fines if returned after due date (0.50 per day)
        fine_amount = 0.00
        if return_date > due_date:
            days_late = (return_date - due_date).days
            fine_amount = round(days_late * 0.50, 2)

        branch = random.choice(branches)
        patron = random.choice(patron_types)
        cost = random.choice(categories[category_name]["cost"])

        writer.writerow([
            transaction_id,
            checkout_date.isoformat(),
            due_date.isoformat(),
            return_date.isoformat(),
            branch,
            patron,
            category_name,
            f"{fine_amount:.2f}",
            f"{cost:.2f}"
        ])

print(f"Successfully generated {TOTAL_ROWS} rows in {FILENAME}")
```
